```python
from optimization_logic import *

import pandas as pd
import matplotlib.pyplot as plt
from datetime import timedelta
import os

import warnings
warnings.filterwarnings("ignore")

#Plotting styles
# plt.style.use('ggplot')
plt.rcParams.update(
{"figure.figsize": (12, 5),
#  'font.family': 'serif',
"text.usetex": False,
"axes.titlesize" : 12,
"axes.labelsize" : 10,
# "axes.labelweight" : "bold",
"lines.linewidth": 1,
"lines.markersize" : 5,
"legend.fontsize": 8,
"legend.title_fontsize" : 9,
"xtick.labelsize" : 8,
"ytick.labelsize" : 8})
```

# Parameters

```python
"""
SIZE: maximal power content of the battery. Units have to be chosen here and be consistent with other power/energy related attributes
POWER_UP: maximal power for discharging; units are analogous to units as size (e.g. kW -> kWh, MW -> MWh)
POWER_DOWN: maximal power for charging; units are analogous to units as size (e.g. kW -> kWh, MW -> MWh)
CHARGE_EFFICIENCY: efficiency factor to compute SoC change when charging the asset, fraction between [0,1]
DISCHARGE_EFFICIENCY: efficiency factor to compute SoC change when discharging the asset, fraction between [0,1]

INIT_SOC: initial energy content of the battery at the start of simulation period (in energy units). (default = None -> size/2 = 50%)
END_SOC: energy content of the battery at end of each day (in energy units). (default = None -> no such constraint)
# the energy content at the start of each day (after the very first day) is taken as the end_SOC for previous day

end_soc_limit_type: the kind of contraint applied for end of day SOC. Chose between ['equal_to', 'greater_than_or_equal_to', 'less_than_or_equal_to']. 
                    Applied only when END_SOC != None

MANUAL_SOC_MIN = minimum allowed SOC
MANUAL_SOC_MAX = maximum allowed SOC

MAX_THROUGHPUT_CYCLES: maximum throughput/discharge cycles allowed in 1 day
THROUGHPUT_SIZE: throughput size (MWh) upon which a 'cycle' is defined, daily discharge throughput <= MAX_THROUGHPUT_CYCLES * THROUGHPUT_SIZE MWh
TSTEP: fraction of an hour (1/N), where N denotes the frequency of settlements made per hour

SOLUTION_TIME_LIMIT_SEC = max time allowed for the optimization algo to find a solution (in sec.)
INTEGER_PN = True if the setpoints (MW) can only be integer values, else False

assume_perfect_foresight = True if perfect foresight on DA prices is to be assumed, otherwise False
                           When False, the optimization will be done on DA forecast, and settlement on actuals
"""

###############################
#       BESS specifics        #
###############################

SIZE = 100                # MWh - default 56.4
POWER_UP = 50            # MW - default 49
POWER_DOWN  = 50         # MW - default 49

RTE = 0.88                                # round trip efficiency
CHARGE_EFFICIENCY = np.sqrt(RTE)          #default = sqrt(RTE)
DISCHARGE_EFFICIENCY = np.sqrt(RTE)       #default = sqrt(RTE)

MAX_THROUGHPUT_CYCLES = 2              # count - default 2.3
THROUGHPUT_SIZE = SIZE                    # MWh
TSTEP = 1                                 # default = 1 for hourly DA market

###############################
#      script technicals      #
###############################

SOLUTION_TIME_LIMIT_SEC = 30  # default 30
INTEGER_PN = False            # default True

###############################
#      DA scpecifics          #
###############################

MANUAL_SOC_MIN = 0           # MWh 
MANUAL_SOC_MAX = SIZE         # MWh

INIT_SOC = 0        # MWh, default = size/2 
END_SOC = 0         # MWh, default = None
end_soc_limit_type = 'equal_to'

###############################
#    Simulation specifics     #
###############################

assume_perfect_foresight = True

###############################
#           Plotting          #
###############################

show_daily_plots = True

###############################
#           Logging           #
###############################

save_setpoint_log = False
save_daily_log = False
save_path = '/simulation_logs/'
save_name_prefix = f'martina_{SIZE}mwh_{POWER_DOWN}mw'


settlement_freq = {0.25: "15min", 0.5: "30min", 1: "H"}
```

# Data

```python
data_files_path = os.getcwd()+'/data_files'
file_name = 'gb_epex_da_actuals.csv'
main_data = pd.read_csv(data_files_path+'/'+file_name)
```

```python
main_data.head()
```

```python
da_data = main_data[["dtutc", "da_price"]].copy()
da_data["dtutc"] = pd.to_datetime(da_data.dtutc, utc=True)
```

```python
# here we resample the dataset to the frequency of DAM, as specified in param 'TSTEP' above
da_data = da_data.resample(settlement_freq[TSTEP], on = "dtutc", origin="start").mean()
da_data
```

```python
da_data.plot()
```

```python
# here we add a 'date' column
da_data.reset_index(inplace=True)
da_data["dtcet"] = da_data.dtutc.dt.tz_convert("Europe/Brussels")
da_data["date"] = da_data.dtcet.dt.date
```

# Simulations

```python
simulation_start_date = pd.to_datetime("2024-07-31").date()
simulation_stop_date = pd.to_datetime("2024-10-31").date()
```

```python
dates = da_data[(da_data.date>=simulation_start_date)&(da_data.date<=simulation_stop_date)].date.unique()
dates
```

```python
# initiate class object, once at the start of the simulation

# initiate class object
bess_object_dam = day_ahead(
    size=SIZE,
    power_up=POWER_UP,
    power_down=POWER_DOWN,
    timestep=1,
    charge_eff=CHARGE_EFFICIENCY,
    discharge_eff=DISCHARGE_EFFICIENCY,
    manual_soc_max=MANUAL_SOC_MAX,
    manual_soc_min=MANUAL_SOC_MIN,
    init_soc=INIT_SOC,
    end_soc=END_SOC,
    end_soc_limit_type=end_soc_limit_type,
    daily_cycle_max=MAX_THROUGHPUT_CYCLES,
    throughput_size=THROUGHPUT_SIZE,
    solution_time_limit=SOLUTION_TIME_LIMIT_SEC,
    integer_solution=INTEGER_PN,
)
```

```python
log = pd.DataFrame()
log_setpoint = pd.DataFrame()
# today_starting_soc = init_soc

for d in dates[:]:

    # ref_date = # for d, from your own dataset 

    # actuals, for price settlements
    today_da_data = da_data[da_data.date == d].reset_index(drop=True)

    # forecast for scheduling
    
    DAM_run = bess_object_dam.solving(
        dam_actual=today_da_data.da_price,
        dam_fc=None #for perfect foresight
        plot=show_daily_plots,  # toggle plotting here
        title_text1="DA Auction 1",
        title_text2=f"{d}",
    )
    DAM_run["dtutc"] = today_da_data.dtutc
    DAM_run["da_actual"] = today_da_data.da_price
    eod_soc = DAM_run.soc_end.iloc[-1]

    # logging
    log_setpoint = pd.concat(
        [
            log_setpoint,
            DAM_run[
                ["net", "soc_start", "soc_end", "dtutc", "da_actual"]
            ],
        ],
        ignore_index=True,
    )
    log = pd.concat(
        [
            log,
            pd.DataFrame(
                [
                    {
                        "date": d,
                        "pnl_da": bess_object_dam.pnl,
                        "charge_mwh": bess_object_dam.charge_mwh,
                        "discharge_mwh": bess_object_dam.discharge_mwh,
                        "init_soc": bess_object_dam.init_soc,
                        "end_soc": eod_soc,
                        "cycles": bess_object_dam.cycles,
                    }
                ]
            ),
        ],
        ignore_index=True,
    )

    bess_object_dam.init_soc = eod_soc
    print(log.tail(1))
```

```python
if save_setpoint_log:
    log_setpoint.to_csv(os.getcwd()+save_path+save_name_prefix+"_setpoint_log.csv", index=False)

if save_daily_log:
    log.to_csv(os.getcwd()+save_path+save_name_prefix+"_daily_log.csv", index=False)

```

```python
def get_monthly_figures(daily_log):
    log_monthly = daily_log.copy()
    log_monthly['date'] = pd.to_datetime(log_monthly.date)
    log_monthly = log_monthly.resample('M', on = 'date').mean()

    log_monthly.loc[pd.to_datetime(daily_log.loc[0, 'date']), :] = np.nan
    log_monthly.sort_index(inplace=True)
    log_monthly.bfill(inplace=True)
    return log_monthly
```

```python
log_monthly = get_monthly_figures(daily_log=log.copy())
log_monthly
```

# Visualization

```python
def plot_pl_distribution(daily_log):
         
    plt.hist(daily_log['pnl_da'], bins=50, edgecolor='black')
    plt.xlim((0))
    plt.title(f"Daily P&L distribution ({POWER_DOWN}MW/{SIZE}MWh)\n Backtesting Period: {daily_log.date.iloc[0]} --> {daily_log.date.iloc[-1]} ({len(daily_log)} days)")
    plt.xlabel("EUR")
    plt.ylabel("Counts")
    plt.show()
    return

def plot_overview(daily_log, monthly_log):
    
    fig, ax = plt.subplots(2,2, sharex=True, figsize=(15,6))

    ax[0,0].plot(daily_log.date, daily_log.pnl_da.cumsum(), label = f"Cumulative PnL\n(Total = {round(sum(daily_log.pnl_da),2)})")
    ax[1,1].plot(da_data.groupby('date').max().da_price-da_data.groupby('date').min().da_price, label = "daily spread")
    # ax[1,0].plot(log.date, log.charge_mwh/log.discharge_mwh, c = 'orange', label = "daily Charge_MWh:Discharge_MWh")

    ax[0,1].plot(daily_log.date, daily_log.pnl_da)
    ax[0,1].step(monthly_log.index,
                monthly_log.pnl_da,
                    c='k', ls='--', label = 'monthly avg.') 
    # ax[0,1].axhline(np.mean(log.pnl_da), c = 'k', ls = '--', label = f"avg. = {round(np.mean(log.pnl_da), 2)}")

    ax[1,0].plot(daily_log.date, daily_log.cycles)
    ax[1,0].axhline(np.mean(daily_log.cycles), c = 'k', ls = '--', label = f"avg. = {round(np.mean(daily_log.cycles), 2)}")

    ax[1,0].legend(title = "Daily Cycles")
    ax[0,1].legend(title = "Daily PnL")
    ax[1,1].legend()
    ax[0,0].legend()

    ax[1,1].grid()
    ax[0,1].grid()
    ax[1,0].grid()
    ax[0,0].grid()

    ax[0,0].set(ylabel = "EUR")
    ax[0,1].set(ylabel = "EUR")
    ax[1,0].set(ylabel = "count")
    ax[1,1].set(ylabel = "EUR/MWh")

    plt.suptitle(f"Backtesting Period: {daily_log.date.iloc[0]} --> {daily_log.date.iloc[-1]} ({len(daily_log)} days)")
    plt.tight_layout()
    plt.show()
    return

def print_metrics(log_file):
    print(f"Backtesting Period: {log_file.date.iloc[0]} --> {log_file.date.iloc[-1]} ({len(log_file)} days)")
    print("\n------------------------------------------------------------------\n")
    print(f"Gross P&L (EUR): {round(log_file.pnl_da.sum(),2)}")
    print(f"Gross Throughput (MWh): {round(log_file.discharge_mwh.sum(),2)}")
    print(f"Gross Throughput Cycles (count): {round(log_file.cycles.sum(),2)}")
    print(f"P&L/MWh discharged (EUR/MWh): {round(log_file.pnl_da.sum()/log_file.discharge_mwh.sum(),2)}")
    print(f"Annualised P&L (kEUR/MW/year): {round(log_file.pnl_da.sum()*365/(len(log_file)*POWER_DOWN)/1000,2)}")

    print("\n------------------------------------------------------------------\n")
    print(f"Average daily P&L (EUR): {round(log_file.pnl_da.mean(),2)}")
    print(f"Average daily Charge (MWh): {round(log_file.charge_mwh.mean(),2)}")
    print(f"Average daily Discharge (MWh): {round(log_file.discharge_mwh.mean(),2)}")
    print(f"Average daily Throughput Cycles (count): {round(log_file.cycles.mean(),2)}")
    
    return
```
