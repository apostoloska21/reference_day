```python
from optimization_logic import *
from benchmark import *
import pandas as pd
import matplotlib.pyplot as plt
from datetime import timedelta
import warnings

warnings.filterwarnings("ignore")
```

```python
data = pd.read_csv('../src/data/processed/gb_epex_da_actuals.csv', parse_dates=['dtutc', 'dtcet'])
print(data.head())
```

```python
benchmark_daily, benchmark_total = benchmark_model(data)
print(f"Benchmark Total PnL: {benchmark_total}")
```

```python
import matplotlib.pyplot as plt

daily_revenues_df = pd.DataFrame(list(benchmark_daily.items()), columns=['date', 'revenue'])

daily_revenues_df['date'] = pd.to_datetime(daily_revenues_df['date'])

plt.figure(figsize=(12, 6))
plt.plot(daily_revenues_df['date'], daily_revenues_df['revenue'], marker='o', label='Daily PnL')
plt.axhline(0, color='red', linestyle='--', alpha=0.7)
plt.xlabel('Date')
plt.ylabel('PnL (GBP)')
plt.title('Daily Profit and Loss (PnL)')
plt.legend()
plt.grid(alpha=0.5)
plt.show()
```
